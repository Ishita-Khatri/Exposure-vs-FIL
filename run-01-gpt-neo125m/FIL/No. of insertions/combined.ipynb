```python
import wandb
import pandas as pd

entity = "ishita49-indian-institute-of-technology-madras"
project = "3_FIL - No shuffle"
api = wandb.Api()

runs = api.runs(f"{entity}/{project}")

fil_runs = [run for run in runs if "fil" in run.tags]
base_fil = [run for run in runs if "base:857" in run.tags]
base_fil_413 = [run for run in runs if "base:413" in run.tags]
acc_runs = [run for run in runs if not run.name.startswith("FIL")]

for run in fil_runs:
    print(run.name)
print('\n')
for run in base_fil:
    print(run.name)
print('\n')
for run in base_fil_413:
    print(run.name)
print('\n')
for run in acc_runs:
    print(run.name)
```

```python

for fil_run, acc_run in zip(fil_runs, acc_runs):
    fil_df = fil_run.history()
    acc_df = acc_run.history()

    final_fil = fil_df["FIL"].dropna().iloc[-1]
    final_acc = acc_df["Validation accuracy"].dropna().iloc[-1]

    fil_run.summary["Final FIL"] = final_fil
    fil_run.summary["final_accuracy"] = final_acc
    fil_run.update()
```

```python
for fil_run, base_fill, acc_run in zip(fil_runs, base_fil, acc_runs):
    fil_df = fil_run.history()
    base_df = base_fill.history()
    acc_df = acc_run.history()

    run = wandb.init(
        project=project,
        name=f'Diff {fil_run.name}',
        entity=entity,
        tags=acc_run.tags + ['diff'] 
    )

    final_acc = acc_df["Validation accuracy"].dropna().iloc[-1]

    final_diff = None
    for step, (fil, base) in enumerate(zip(fil_df["FIL"], base_df["Baseline FIL"])):
        diff = fil - base
        final_diff = diff
        wandb.log({"diff": diff, "example": step}, step=step)

    run.summary["Final FIL (Diff)"] = final_diff
    run.summary["final_accuracy"] = final_acc
    wandb.finish()
```

```python
for fil_run, exp_run in zip(fil_runs, exp_runs):
    fil_df = fil_run.history()
    exp_df = exp_run.history()
    parts = exp_run.name.split(",")
    num = int(parts[0].split("=")[1])
    epsilon = int(parts[1].split("=")[1])
    run = wandb.init(
        project='Comparison - Exposure vs FIL',
        name=exp_run.name,
        entity=entity)
    fil = fil_df["FIL"].dropna().iloc[-1]
    exp = exp_df["Exposure"].dropna().iloc[-1]

    wandb.log({'Exposure':exp,'FIL':fil})
    wandb.finish()
```

```python
import wandb

entity = "ishita49-indian-institute-of-technology-madras"
project_fil = "3_FIL - No. of insertions"
project_exp = "2_DPSGD - No of insertions"

api = wandb.Api()

runs_fil = api.runs(f"{entity}/{project_fil}")
runs_exp = api.runs(f"{entity}/{project_exp}")

name_fil = "FIL,epsilon=2"
name_exp = "Exposure, Epsilon=2"

target_run_fil = next((run for run in runs_fil if run.name == name_fil), None)
target_run_exp = next((run for run in runs_exp if run.name == name_exp), None)
```

```python
fil_df = target_run_fil.history()
exp_df = target_run_exp.history()
print(fil_df)
print(exp_df)
```

```python
for i in range(7):
    run = wandb.init(
        project='2_Comparison - Exposure vs FIL (number of insertions)',
        name=f'Num={2**i}',
        entity="ishita49-indian-institute-of-technology-madras"
    )
    
    # Log one point
    wandb.log({
        'FIL': fil_df["FIL"].iloc[i],
        'Exposure': exp_df["Exposure"].iloc[i + 1],
        'Number of insertions':2**i
    })

    run.finish()
```
